## 2. Phạm vi dữ liệu

Dự án sử dụng toàn bộ dataset TiniX Vietnamese Job Descriptions. Theo yêu cầu mới, dữ liệu được chia thành 90% train và 10% test. Tập test không trùng với tập train.

Trong bài phân cụm, tập train được dùng để fit các bước xử lý đặc trưng và mô hình phân cụm. Tập test được dùng để kiểm tra khả năng gán cụm cho dữ liệu mới và demo 10 mẫu test.

## 3. Nhóm đặc trưng sử dụng

| Nhóm | Cột dự kiến | Vai trò |
|---|---|---|
| Mô tả công việc | `job_title`, `job_description` | Hiểu bản chất công việc |
| Yêu cầu ứng viên | `requirements`, `experience_level`, `education_level` | Hiểu kỹ năng, kinh nghiệm, trình độ |
| Quyền lợi | `benefits` | Bổ sung thông tin phúc lợi |
| Đặc điểm tuyển dụng | `salary`, `location`, `job_type`, `job_industry`, `job_position`, `company_name`, `year` | Bổ sung bối cảnh tuyển dụng |

## 4. Input và output

### Input

Một tin tuyển dụng gồm tiêu đề, mô tả, yêu cầu, quyền lợi, lương, địa điểm, hình thức làm việc, ngành nghề, vị trí, kinh nghiệm, trình độ học vấn và năm đăng tin.

### Output

Output là `cluster_id`. Sau khi có `cluster_id`, nhóm sẽ diễn giải từng cụm bằng top keywords, ngành nghề phổ biến, lương, địa điểm và yêu cầu kinh nghiệm.

## 5. Câu hỏi nghiên cứu

1. Có thể chia các tin tuyển dụng thành bao nhiêu nhóm chính?
2. Mỗi cụm đại diện cho nhóm công việc hoặc phân khúc tuyển dụng nào?
3. Các từ khóa đặc trưng của từng cụm là gì?
4. Cụm nào có mức lương trung vị cao hơn?
5. Cụm nào yêu cầu kinh nghiệm hoặc trình độ học vấn cao hơn?
6. Cụm nào tập trung nhiều ở Hà Nội, TP.HCM hoặc các địa phương khác?
7. Việc kết hợp đặc trưng văn bản và đặc trưng có cấu trúc có giúp tạo ra các cụm dễ diễn giải hơn không?

## 6. Pipeline dự kiến

```text
Raw data
→ Train/test split 90/10
→ Data cleaning
→ Text cleaning
→ Salary parsing
→ TF-IDF
→ TruncatedSVD / LSA
→ Structured feature encoding
→ Clustering
→ Cluster evaluation
→ Cluster interpretation
→ Demo 10 mẫu test
```
